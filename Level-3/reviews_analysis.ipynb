```python
import pandas as pd
import matplotlib.pyplot as plt
import os

os.makedirs("screenshots", exist_ok=True)

# Load dataset
df = pd.read_csv('../Dataset/dataset.csv')

# Drop missing ratings
df = df[df['Aggregate rating'].notna()]

# Highest & lowest vote restaurants
max_votes = df[df['Votes'] == df['Votes'].max()]
min_votes = df[df['Votes'] == df['Votes'].min()]

print("Restaurant(s) with Highest Votes:\n", max_votes[['Restaurant Name', 'Votes', 'Aggregate rating']])
print("\nRestaurant(s) with Lowest Votes:\n", min_votes[['Restaurant Name', 'Votes', 'Aggregate rating']])

# Correlation between Votes and Rating
plt.scatter(df['Votes'], df['Aggregate rating'], alpha=0.5)
plt.title("Votes vs Rating")
plt.xlabel("Number of Votes")
plt.ylabel("Aggregate Rating")
plt.savefig("screenshots/votes_vs_rating.png")
plt.show()

# Optional: Distribution of Votes
plt.hist(df['Votes'], bins=30, color='skyblue')
plt.title("Votes Distribution")
plt.xlabel("Votes")
plt.ylabel("Number of Restaurants")
plt.savefig("screenshots/votes_distribution.png")
plt.show()
```
