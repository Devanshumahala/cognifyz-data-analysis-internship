```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Screenshot folder safety
os.makedirs("screenshots", exist_ok=True)

# Load dataset
df = pd.read_csv('../Dataset/dataset.csv')

# Price range count
price_counts = df['Price range'].value_counts().sort_index()
print("Price Range Counts:\n", price_counts)

# Percentages
percent_price = (price_counts / len(df)) * 100
print("\nPercentages:\n", percent_price)

# Plot
price_counts.plot(kind='bar', color='green')
plt.title("Price Range Distribution")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.tight_layout()
plt.savefig("screenshots/price_range.png")
plt.show()
```
