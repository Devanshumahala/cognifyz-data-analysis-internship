```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Screenshot folder bana lo (safety)
os.makedirs("screenshots", exist_ok=True)

# Load dataset
df = pd.read_csv('../Dataset/dataset.csv')

# Clean cuisines
df = df[df['Cuisines'].notna()]
df['Main_Cuisine'] = df['Cuisines'].apply(lambda x: x.split(',')[0].strip())

# Top 3 cuisines
top_cuisines = df['Main_Cuisine'].value_counts().head(3)
print("Top 3 Cuisines:\n", top_cuisines)

# Percentages
percent = (top_cuisines / len(df)) * 100
print("\nPercentages:\n", percent)

# Plot
top_cuisines.plot(kind='bar', color='orange')
plt.title("Top 3 Cuisines")
plt.ylabel("Number of Restaurants")
plt.xlabel("Cuisine")
plt.tight_layout()
plt.savefig("screenshots/top_cuisines.png")
plt.show()
```
